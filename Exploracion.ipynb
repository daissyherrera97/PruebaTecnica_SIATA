## Cargar información

```python
data_path_367= 'Datos/Estacion_meteorologica_367_2025-12-01_2026-04-30.csv'#Meteorológica vaisala
data_path_803= 'Datos/Estacion_nivel_803_2025-12-01_2026-04-30.csv'#Nivel
data_path_6004= 'Datos/Estacion_piranometro_6004_2025-12-01_2026-04-30.csv'#Piranómetro
data_path_35= 'Datos/Estacion_pluviometrica_35_2025-12-01_2026-04-30.csv'#Pluviómetro
```

```python
def lectura_csv(path):
    data=pd.read_csv(path, sep=',',header=0)
    return data
```

```python
data_367=lectura_csv(data_path_367)
data_803=lectura_csv(data_path_803)
data_6004=lectura_csv(data_path_6004)
data_35=lectura_csv(data_path_35)
```
